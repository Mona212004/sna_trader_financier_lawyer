```python
# clustering on trained data
import pandas as pd
import dedupe
import os

# `partition` will return sets of records that dedupe believes are all referring to the same entity.

# format df123 into dict struct {row_id: {field: value}}
df123 = pd.read_csv(r"D:\commo\code\entity_resolution\df123_train.csv")
data_d = {}
for idx, row in df123.iterrows():
    if isinstance(row["name"], str) and row["name"].strip():
        data_d[idx] = {"name": row["name"]}

# If a settings file already exists, we'll just load that and skip training
settings_file = r"D:\commo\code\entity_resolution\dedupe_settings"
if os.path.exists(settings_file):
    print("reading from", settings_file)
    with open(settings_file, "rb") as f:
        deduper = dedupe.StaticDedupe(f)
else:
    raise FileNotFoundError(f"No trained settings found at {settings_file}. Run the training script first.")

print("clustering...")
# clustered_dupes = deduper.partition(data_d, 0.5) #284 duplicate sets
clustered_dupes = deduper.partition(data_d, 0.5)  
#clustered_dupes = deduper.partition(data_d, 0.9)  # 349 duplicate sets

print("# duplicate sets", len(clustered_dupes))
```

```python
# ## Writing Results
import csv
# Write our original data back out to a CSV with a new column called
# 'Cluster ID' which indicates which records refer to each other.

cluster_membership = {}
for cluster_id, (records, scores) in enumerate(clustered_dupes):
    for record_id, score in zip(records, scores):
        cluster_membership[record_id] = {
            "Cluster ID": cluster_id,
            "confidence_score": score,
        }

df123['Cluster ID'] = df123.index.map(lambda i: cluster_membership.get(i, {}).get('Cluster ID'))
df123['Match Confidence'] = df123.index.map(lambda i: cluster_membership.get(i, {}).get('confidence_score'))

output_file = r"D:\commo\code\entity_resolution\clustering_output.csv"
df123.to_csv(output_file, index=False)
print(f"saved to {output_file}")
```
