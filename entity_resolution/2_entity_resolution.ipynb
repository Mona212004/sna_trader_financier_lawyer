```python
# split the df2 Known Clients into its own pd.series
def split_comma_list(cell):
    if not isinstance(cell, str) or not cell.strip():
        return []
    parts = re.split(r",\s*(?![^()]*\))", cell)
    return [p.strip() for p in parts]

# split each row's comma-separated string into a list
split_series = df2b.dropna().apply(split_comma_list)

# explode: each item in the list becomes its own row
exploded = split_series.explode().reset_index(drop=True)

# turn into a DataFrame with one column, one entity name per row
df2_clients_df = pd.DataFrame({"Firm Name": exploded})
df2_clients_df 
```
