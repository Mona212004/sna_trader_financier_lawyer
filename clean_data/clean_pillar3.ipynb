```python
# check commo focus
import re

# make a copy
if "Commodity Focus copy" not in p3.columns:
    p3["Commodity Focus copy"] = p3["Commodity Focus"].copy()

# remove brackets
copy = p3['Commodity Focus copy'].str.replace(r'\(.*?\)','', regex=True).str.strip()
# remove double spacing
copy = p3['Commodity Focus copy'].str.replace(r'\s+',' ', regex=True).str.strip()

# 'hard/metals': hard commo (metals), metal, Metals & Mining (base metals, precious metals), copper, aluminum, metals, base metals, precious metals, critic metals, Metals (base metals, precious metals), Ferrous & Non-Ferrous Metals, Base & Precious Metals (copper, aluminum, zinc, gold), Metals & Minerals, base/precious metals, coal, iron ore, base metals industrial metals, and precious metals, and metals, metals & minerals,
# 'soft/agricultural': soft commo, soft, agribusiness (agricultural materals, soft commodities), agriculture, agribusiness, agricultural materials, soft commodities, Agriculture (agricultural materials, soft commodities). , and Agribusiness (wheat, barley, cotton, sugar, soy, palm oil), and agricultural commodities, Food & Agribusiness, agricultural products, soft commodities, forest products (pulp, paper), bulk physical commodities like cocoa, coffee, grains, rice, soft commodities (biofuels, agricultural),  agricultural products, agriculture/softs
# 'energy': energy, and Energy (petroleum, oil), oil, gas, petroleum, energy transition commodities, Energy (oil, petroleum), energy/oil commodities, Energy (crude oil, fuel oil, LNG), crude oil, petroleum products (major crude oil distillates), iron ore, fuel oil, gas oil, crude oil, oil products, and US/European natural gas, oil, natural gas, power, European power, gas, coal, and North American gas/power, electricity, liquefied natural gas, Energy/oil, general, energy(oil, gas, electricity), Energy (incl. coal, selectively), Energy (biofuels), Energy commo
# 'others': OCBC also covers commodities in both vanillas and derivatives, hedging products (FX, price, interest rate derivatives), General commodity trade financing, fixed income, currencies & commodities services under corporate and institutional banking, specialize in trade banking, corporate banking, investment banking, and international banking services, foreign exchange and fixed income, fx and fixed income, trade finance for a broad range of cross‑border physical commodity trades executed on Zall / CIC B2B platforms
# 'petrochemicals': Chemicals,  petrochemical, polymer,  chemicals & plastics
# 'emerging': sustainable commodities, decarbonisation commodities, and carbon markets
# 'Unknown': n/a; , product agnostic, agnostic, sector‑agnostic SME lender

# ['energy', hard/metals']: energy and hard commo, Metals and Oil, Energy/hard commodity primarily coal trading, hard commodities – non‑ferrous metals, ferro‑alloys, minerals, industrial raw materials and related energy products
# ['energy', hard/metals', 'soft/agriculture']: primarily energy and hard commodities – oil and refined products (crude, fuel oil, middle distillates), gas, coal, emissions, plus metals and some agricultural products
# ['hard/metals','soft/agricultural']: Hard & Soft Commodities

# idk: carbon and trade commodities, conducted trade finance for dairy products, emissions, refined products, refinery products, textiles & apparel, industrial goods & machinery, handicrafts, home furnishings & consumer goods, apparel
# delete: freight, in line with the Traxys group mandate

import re

if "Commodity Focus copy" not in p3.columns:
    p3["Commodity Focus copy"] = p3["Commodity Focus"].copy()

category_patterns = {
    "energy": [
        r"\blng\b", r"liqu[ei]fied natural gas", r"\bcrude\b", r"fuel oil",
        r"natural gas", r"\bcoal\b", r"\bpower\b", r"electricity", r"gas oil",
        r"biofuels?", r"refined products?", r"refinery products?",
        r"middle distillates?", r"\boil\b", r"\bpetroleum\b", r"\bgas\b",
        r"energy transition", r"\benergy\b",
    ],
    "hard/metals": [
        r"base metals?", r"precious metals?", r"industrial metals?",
        r"ferrous", r"non-ferrous", r"ferro-alloys?", r"\bcopper\b", r"\baluminum\b",
        r"\bzinc\b", r"\bgold\b", r"iron ore", r"critic\w* metals?",
        r"\bminerals?\b", r"industrial raw materials?", r"\bmining\b",
        r"\bmetals?\b", r"hard commo\w*",
    ],
    "soft/agricultural": [
        r"agribusiness", r"agricultur\w*", r"soft commodit\w*", r"\bsofts?\b",
        r"\bwheat\b", r"\bbarley\b", r"\bcotton\b", r"\bsugar\b", r"\bsoy\b",
        r"palm oil", r"\bcocoa\b", r"\bcoffee\b", r"\bgrains?\b", r"\brice\b",
        r"\bdairy\b", r"forest products?", r"\bpulp\b", r"\bpaper\b", r"\bfood\b",
    ],
    "petrochemicals": [
        r"petrochemical", r"\bpolymer\b", r"\bplastics?\b", r"\bchemicals?\b",
    ],
    "emerging": [
        r"sustainable commodit\w*", r"decarbonisation", r"decarbonization",
        r"carbon markets?", r"emissions?", r"\bcarbon\b",
    ],
}

def expand_slash_compounds(text):
    return re.sub(
        r"\b([\w-]+)\s*(?:/|&|,?\s+and)\s*([\w-]+)\s+(metals?|commodities|natural gas|banking)\b",
        r"\1 \3, \2 \3", text, flags=re.IGNORECASE,
    )

def classify_row(raw):
    if not isinstance(raw, str) or not raw.strip():
        return ["Unknown"]

    text = re.sub(r"\s*\(([^()]*)\)\s*$", "", raw).strip()  # drop trailing (source citation)

    # only a bare "n/a" / "n/a;" with nothing else -> Unknown
    if re.match(r"^n/a\s*;?\s*$", text, flags=re.IGNORECASE):
        return ["Unknown"]
    if re.search(r"\bagnostic\b", text, flags=re.IGNORECASE):
        return ["Unknown"]

    text_l = expand_slash_compounds(text).lower()

    matched = [cat for cat, patterns in category_patterns.items()
               if any(re.search(p, text_l) for p in patterns)]

    return matched if matched else ["others"]

p3["Commodity Focus"] = copy.apply(classify_row)
p3["Commodity Focus"] = p3["Commodity Focus"].apply(
    lambda x: ", ".join(x) if isinstance(x, list) else x
)
p3["Commodity Focus"]
```

```python
# check 'Known Clients'
p3["Known Clients"].unique()

# delete the sources because they are all perplexity and the only one from an official source don't mention any names. All are officially undisclosed.
# any row that includes 'n/a' or nan > replace the whole row with 'Unkown'
#  'SMEs in bulk commodity markets (https://tradeflow.capital/esg)' = no names = replace with 'Unkown'
# 'P.T Kebun Tebu Mas (KTM), commodity producers, processors, and buyers along its supply chain (e.g., sugar mills, food ingredient suppliers, energy product buyers) (perplexity)' = replace with 'P.T Kebun Tebu Mas (KTM)
# keep brackets but no split if inside brackets
#'SMEs (annual revenues typically below USD 50m); Exporters of plastic products, frozen foods,  red chillies, Basmati rice, shrimp and rice, avocados, Textiles and industrial equipment  (perplexity)' = no names = replace with 'Unknown'
# 'Fashion Tex Asia Ltd, Sadat Apparels Ltd, SafeFlex Limited International, Pan-European metal trading firm, US-based apparel trading company, China-based apparel manufacturer, China-based apparel manufacturer (perplexity)' =  'Fashion Tex Asia Ltd, Sadat Apparels Ltd, SafeFlex Limited International'
# Limited or ltd or ltd. becomes Ltd
# pte. becomes pte
# all names have to be capitalized the first letter
#'A Singapore-based marine-sector supplier, A leading bottled-drink manufacturer in Thailand (perplexity)' = 'Unknown'
#'PT Bening Big Tree Farms (BTFI), ACLF portfolio SMEs, Square Yards (India),  India Cements, Bangkok Mass Transit System, Rizal Commercial Banking Corporation (perplexity)' = 'PT Bening Big Tree Farms (BTFI), ACLF portfolio SMEs, Square Yards (India), Rizal Commercial Banking Corporation (perplexity)'
# 'InMobi (India‑based adtech firm) received US$350m loan (perplexity)' = 'Unknown'
#  'a Swiss commodities trading firm active in biofuels (perplexity)' = 'Unknown'
# 'The Prax Group (USD 350 million inventory financing facility), various energy/clean-fuels structures (e.g. Green Plains 45Z credits), and equity/project investments such as Nostromo (perplexity)' =  'The Prax Group (perplexity)'
# GP, SSL, COFCO, LDC, IFC, GTLP, APRIL, UK, BV, ABN AMRO, BNP, CA, TOTSA, ING, MUFG, C2G2, PT , PLC  ,
# IncuBaker, MoneySmart, xSquare
import re

def normalize_key(s):
    if not isinstance(s, str):
        return s
    return (s.replace('\u2011', '-').replace('\u2013', '-')
             .replace('\u2019', "'").strip())

manual_overrides = {
    normalize_key('n/a; Traders, Producers, Processors, Refiners, Distributors, and Corporates with commodity financing needs (perplexity)'): ['Unknown'],
    normalize_key('SMEs in bulk commodity markets (https://tradeflow.capital/esg)'): ['Unknown'],
    normalize_key('P.T Kebun Tebu Mas (KTM), commodity producers, processors, and buyers along its supply chain (e.g., sugar mills, food ingredient suppliers, energy product buyers) (perplexity)'):
        ['P.T Kebun Tebu Mas'],
    normalize_key('n/a; banks, proprietary trading firms, hedge funds, pension funds, asset managers, and large corporations seeking execution/clearing, liquidity, and prime brokerage. (perplexity)'): ['Unknown'],
    normalize_key('SMEs (annual revenues typically below USD 50m); Exporters of plastic products, frozen foods,  red chillies, Basmati rice, shrimp and rice, avocados, Textiles and industrial equipment  (perplexity)'): ['Unknown'],
    normalize_key('Fashion Tex Asia Ltd, Sadat Apparels Ltd, SafeFlex Limited International, Pan-European metal trading firm, US-based apparel trading company, China-based apparel manufacturer, China-based apparel manufacturer (perplexity)'):
        ['Fashion Tex Asia Ltd', 'Sadat Apparels Ltd', 'SafeFlex Limited International'],
    normalize_key('A Singapore-based marine-sector supplier, A leading bottled-drink manufacturer in Thailand (perplexity)'): ['Unknown'],
    normalize_key('PT Bening Big Tree Farms (BTFI), ACLF portfolio SMEs, Square Yards (India),  India Cements, Bangkok Mass Transit System, Rizal Commercial Banking Corporation (perplexity)'):
        ['PT Bening Big Tree Farms', 'Square Yards', 'Rizal Commercial Banking Corporation'],
    normalize_key('InMobi (India\u2011based adtech firm) received US$350m loan (perplexity)'): ['Unknown'],
    normalize_key('a Swiss commodities trading firm active in biofuels (perplexity)'): ['Unknown'],
    normalize_key('The Prax Group (USD 350 million inventory financing facility), various energy/clean-fuels structures (e.g. Green Plains 45Z credits), and equity/project investments such as Nostromo (perplexity)'):
        ['The Prax Group'],
    normalize_key('The bank primarily serves Australian and New Zealand businesses operating in Asia (perplexity)'): ['Unknown'],
    normalize_key('InMobi (India‑based adtech firm) received US$350m loan (perplexity)'):['Unknown'],
    normalize_key('The Prax Group (USD 350 million inventory financing facility), various energy/clean-fuels structures (e.g. Green Plains 45Z credits), and equity/project investments such as Nostromo (perplexity)'): ['The Prax Group']

}

# --- Acronyms that must always render in full uppercase, regardless of source casing ---
ACRONYMS = ['ABN AMRO', 'GP', 'SSL', 'COFCO', 'LDC', 'IFC', 'GTLP', 'APRIL',
            'UK', 'BV', 'BNP', 'CA', 'TOTSA', 'ING', 'MUFG', 'C2G2', 'PT', 'PLC']

def force_acronym_case(text):
    for acr in ACRONYMS:
        pattern = r'\b' + re.escape(acr) + r'\b'
        text = re.sub(pattern, acr, text, flags=re.IGNORECASE)
    return text

# --- Brand names with intentional non-standard casing that must be preserved exactly ---
BRAND_WHITELIST = ['IncuBaker', 'MoneySmart', 'xSquare']

def remove_brackets(text):
    text = re.sub(r'\(.*?\)', '', text)
    return re.sub(r'\s+', ' ', text).strip()

def standardize_suffixes(name):
    name = re.sub(r'\b(limited|ltd)\b\.?,?', 'Ltd', name, flags=re.IGNORECASE)
    name = re.sub(r'\bpte\b\.?,?', 'pte', name, flags=re.IGNORECASE)
    return re.sub(r'\s+', ' ', name).strip()

def capitalize_first(name):
    if not name:
        return name
    if name.isupper():
        return name  # e.g. "ABN AMRO", "ING" alone
    for brand in BRAND_WHITELIST:
        if name.lower().startswith(brand.lower()):
            return brand + name[len(brand):]  # preserve exact brand casing, e.g. "xSquare"
    return name[0].upper() + name[1:]

def is_generic_descriptor(clause):
    c = clause.strip()
    if re.match(r'^(a|an)\s', c, flags=re.IGNORECASE):
        return True
    if re.search(r'\b\w+-based\b', c, flags=re.IGNORECASE):
        return True
    return False

def clean_row(raw):
    if not isinstance(raw, str) or not raw.strip():
        return ['Unknown']

    key = normalize_key(raw)
    if key in manual_overrides:
        return manual_overrides[key]

    text = normalize_key(raw)
    if re.match(r'^n/a\s*;?\s*$', text, flags=re.IGNORECASE):
        return ['Unknown']

    text = remove_brackets(text)
    clauses = [c.strip() for c in text.split(',') if c.strip()]

    kept = []
    for c in clauses:
        if is_generic_descriptor(c):
            continue
        c = standardize_suffixes(c)
        c = force_acronym_case(c)
        c = capitalize_first(c)
        kept.append(c)

    return kept if kept else ['Unknown']

p3["Known Clients"] = p3["Known Clients"].apply(clean_row)
p3["Known Clients"] = p3["Known Clients"].apply(
    lambda x: ", ".join(x) if isinstance(x, list) else x
)
```
