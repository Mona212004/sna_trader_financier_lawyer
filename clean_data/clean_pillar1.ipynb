```python
# update existing tiers based on est. revenue
"""
Reference for the tiers:
https://sbr.com.sg/economy/in-focus/only-1-firms-exceed-100m-enterprises-cluster-below-1m-revenue

Reference for the 4 tiers which is then used to modify into 3 tiers:
- Above S$100 Mil — 4,300 firms (1%) → Tier 1
- Between S$10 Mil and S$100 Mil — 16,500 firms (5%) → Tier 2
- Between S$1 Mil and S$10 Mil — 61,100 firms (17%) → Tier 3
- Less than or equal to S$1 Mil — 276,500 firms (77%) → Tier 3

Current exchange rate (late July 2026): 1 SGD ≈ 0.775 USD

Tiers with USD equivalents:
- Tier 1: Above S$100 Mil (≈ above US$77.5 million)
- Tier 2: S$10 Mil – S$100 Mil (≈ US$7.75 million – US$77.5 million)
- Tier 3: Less than S$10 Mil (≈ less than US$7.75 million)
"""
```

```python
# check 'Commodity Type'
# remove brackets and their contents
p1["Commodity Type copy"] = p1["Commodity Type copy"].str.replace(r"\(.*?\)", "", regex=True)
# remove whitespace and convert to lowercase
p1["Commodity Type copy"] = p1["Commodity Type copy"].str.lower().str.strip()

# update the unique values
commodity_mapping = {
    "energy and petrochemicals commodities  &": "energy and petrochemicals",
    "energy and petrochemicals commodities": "energy and petrochemicals",
    "energy and petrochemical commodities": "energy and petrochemicals",
    "energy - hard commo": "energy or hard/metals",
    "soft commodities": "soft/agricultural",
    "hard commo": "hard/metals",
    "energy commo": "energy",
    "energy and soft commodities": "energy and soft/agricultural",
    "multi": "multi (>2 types)",
}

p1["Commodity Type copy"] = p1["Commodity Type copy"].replace(commodity_mapping)

unique_commo_types = p1["Commodity Type copy"].unique().tolist()
unique_commo_types

# energy or hard/metals become ['energy', 'hard/metals']
# energy and petrochemicals become ['energy', 'petrochemicals']
# energy and soft/agricultural become ['energy', 'soft/agricultural']
import re
def split_commodity_types(commodity_type_str):
    if pd.isna(commodity_type_str):
        return [] 
    elif commodity_type_str in ['multi (>2 types)', 'petrochemicals', 'soft/agricultural', 'emerging', 'hard/metals', 'energy']:
        return [commodity_type_str]
    else:
        parts = re.split(r'\s+and\s+|\s+or\s+', commodity_type_str)
        return [p.strip() for p in parts]

p1['Commodity Type'] = p1['Commodity Type copy'].apply(split_commodity_types) #can run only once, comment out after running

# get the unique values of the updated column where each row is a list of str, update() use with set() to append only unique values
unique_values2 = set()
for list in p1['Commodity Type']:
    unique_values2.update(list)
print(unique_values2)
```

```python
# check Listed / Exchange
# remove brackets & whitespace
p1["Exchange"] = p1["Exchange"].str.strip()

# pattern to find inside each cell, its canoncial name / transformation. Format as list of tuples [(,),(,),(,)...] where (,) has (pattern, cannonical)
# \b*\b is word boundary. Ex: \bLSE\b will extract only LSE and not PULSE.
standardized_exchange = [
    (r"\blse\b|london", "London Stock Exchange LSE"),
    (r"\bnyse\b|new york", "New York Stock Exchange NYSE"),
    (r"\btse\b|tokyo", "Tokyo Stock Exchange TSE"),
    (r"\bset\b|thailand", "The Stock Exchange of Thailand SET"),
    (r"\bams\b|amsterdam", "Euronext Amsterdam AMS"),
    (r"\bparis\b", "Euronext Paris ENX"),
    (r"\bsgx\b|singapore", "Singapore Exchange SGX"),
    (r"\bhkex\b|hong kong", "Stock Exchange of Hong Kong HKEX"),
    (r"\bkrx\b|korea", "Korea Exchange KRX"),
    (r"\bnasdaq\b", "Nasdaq Stock Market NDAQ"),
    (r"\basx\b|australian securities", "Australian Securities Exchange ASX"),
    (r"\bidx\b|indonesia", "Indonesia Stock Exchange IDX"),
    (r"\bmyx\b|malaysia", "Malaysia Exchange MYX"),
    (r"\bsse\b|shanghai", "Shanghai Stock Exchange SSE"),
]

def standardize(text):
    #check if the cell is na or empty str or unknown
    if pd.isna(text) or text.strip() == '' or 'unknown' in text.lower():
        #return it 
        return text
    else:
        text = text.lower() #put here so capitalized 'Unknown' doesn't become lowercase
        #list to store unique exchange 
        found = []
        #text has exchange, find the pattern in the text
        for pattern, cannonical in standardized_exchange:
            if re.search(pattern, text, re.IGNORECASE): #returns true of pattern is found in text
                if cannonical not in found: #avoids duplication of the same cannonical if pattern is found once
                    found.append(cannonical)
    #once all patterns are found, join the cannonicals as a string and return else return original text if patterns were not found
    return ', '.join(found) if found else text

# run this function for every row in the exchange column
p1['Exchange'] = p1['Exchange'].apply(standardize)
p1["Exchange"].unique()
```
