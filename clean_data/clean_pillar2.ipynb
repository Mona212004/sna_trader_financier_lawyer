```python
# check the partners
p2['(SG) Partners Specialized in Commodities / Trade / Shipping'].unique()
# strip whitespace
# change , to ; between Managing Partner: n/a, Partners :Jun Bautista, Kimarie Cheang, Sumeet Malhotra
import re


def normalize_separators(text):
    if pd.isna(text):
        return text
    # Replace a comma that is immediately followed by "Word(s): " (a new position label) with a semicolon
    # This pattern looks for: comma, optional space, then 1-4 capitalized words, then a colon
    text = re.sub(r",\s*(?=[A-Z][a-zA-Z\s]{0,30}:)", "; ", text)
    return text


#p2["Key Contacts"] = p2["Key Contacts"].apply(normalize_separators)

test1 = "Managing Partner: n/a, Partners :Jun Bautista, Kimarie Cheang, Sumeet Malhotra, Merrick White"
print(normalize_separators(test1))
```
