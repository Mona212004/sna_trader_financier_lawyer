```python
# check Practice Areas
# normalize delimiters
p2['Practice Areas'] = (p2['Practice Areas']
 .str.replace(r'\s*;\s*', ', ', regex=True) # \s* >= 0 whitespace
 .str.replace(r',\s*and\s+', ', ', regex=True) # \s+ >0 whitespace. X, and Y -> X, Y
 .str.replace(r'\s+and\s+(?=[A-Za-z])', ', ', regex=True) # (?) is the condition. ' and letters' becomes ', letters'
 .str.replace(r'\s+,\s*', ', ', regex=True) # '  ,...' becomes ', '
 .str.replace(r'\(.*?\)', '', regex=True)
 .str.strip())

# fix common typos
import re 
typo_fixes = {
    r"\bemploymebt\b": "employment",
    r"\bcommerical\b": "commercial",
    r"\bcommerical\b": "commercial",
    r"\bmartime\b": "maritime",
    r"\bconveynacing\b": "conveyancing",
    r"\bshp finance\b": "ship finance",
    r"\bemployement\b": "employment",
    r"\bproerty\b": "property",
    r"\bintl trade\b": "international trade",
    r"\bintl\b": "international",
}
for key, value in typo_fixes.items():
    p2['Practice Areas'] = p2['Practice Areas'].str.replace(key, value, regex=True, flags=re.IGNORECASE)

p2['Practice Areas'].unique()
```

```python
import re

# Standardized categories with keyword triggers (checked via substring match, case-insensitive)
category_keywords = {
    'Corporate & M&A': ['corporate', 'm&a', 'mergers', 'acquisitions', 'corporate/m&a', 'corporate & commercial',
                         'commercial & corporate', 'corporate commercial', 'business establishment',
                         'corporate secretarial', 'corporate services', 'corporate governance', 'commercial'],
    'Banking & Finance': ['banking', 'finance', 'financial services', 'project finance', 'structured finance',
                           'fund finance', 'hybrid capital', 'sustainable finance', 'asset finance', 'ship finance',
                           'securities law', 'financial institutions'],
    'Capital Markets': ['capital markets'],
    'Private Equity & Venture Capital': ['private equity', 'venture capital', 'venture technology'],
    'Funds & Investment Management': ['investment funds', 'funds & investment', 'asset & wealth management',
                                       'fund finance'],
    'Islamic Finance': ['islamic finance'],
    'Dispute Resolution & Litigation': ['dispute', 'litigation', 'litigations'],
    'International Arbitration': ['arbitrator','arbitration', 'mediation', 'mediators'],
    'Restructuring & Insolvency': ['restructuring', 'insolvency', 'bankruptcy'],
    'Fraud, Investigations & White Collar': ['fraud', 'asset recovery', 'investigations', 'white collar', 'crime',
                                              'criminal law'],
    'Regulatory & Compliance': ['regulatory', 'compliance', 'sanctions', 'export controls', 'investigations, compliance',
                                 'financial regulatory'],
    'Competition & Antitrust': ['competition', 'antitrust'],
    'Real Estate & Construction': ['real estate', 'construction', 'conveyancing', 'property law', 'reits',
                                    'property & tenancy'],
    'Employment & Labor': ['employment', 'labour', 'labor', 'pensions', 'immigration', 'workers'],
    'Intellectual Property': ['intellectual property', 'trademark', 'patent'],
    'Tax': ['tax'],
    'Shipping & Maritime': ['shipping', 'maritime', 'admiralty', 'ship build', 'ship repair', 'charterparties'],
    'International Trade & Commodities': ['trade', 'commodities', 'wto', 'customs', 'offshore'],
    'Energy & Infrastructure': ['energy', 'infrastructure', 'natural resources', 'oil & gas', 'renewables',
                                 'power projects', 'epc', 'projects'],
    'Private Wealth & Estate Planning': ['private wealth', 'private client', 'wills', 'probate', 'estate',
                                          'trusts', 'wealth inheritance', 'family offices'],
    'Family Law': ['family law', 'family,'],
    'Technology, Data & Cybersecurity': ['technology', 'data', 'cybersecurity', 'privacy', 'digital', 'fintech',
                                          'blockchain', 'digital assets', 'artificial intelligence', 'ai,'],
    'Insurance': ['insurance', 'reinsurance'],
    'Healthcare & Life Sciences': ['healthcare', 'life sciences', 'medical', 'pharmaceuticals', 'medical negligence'],
    'Aviation': ['aviation'],
    'ESG & Sustainability': ['esg', 'sustainability', 'environmental, social', 'carbon credits', 'environment & climate'],
    'Hospitality & Leisure': ['hospitality', 'leisure', 'gaming', 'gambling', 'sports'],
    'Entertainment & Media': ['entertainment', 'media', 'telecommunications'],
    'Defamation & Reputation': ['defamation', 'reputation'],
}

def standardize_practice_areas(raw_string):
    # split into rough tokens on commas, but keep '&' groupings intact by lowering first
    text = raw_string.lower()
    matched = []
    for category, keywords in category_keywords.items():
        for kw in keywords:
            if kw in text:
                matched.append(category)
                break  # only add category once
    # preserve a stable, de-duplicated order
    seen = set()
    ordered = []
    for m in matched:
        if m not in seen:
            seen.add(m)
            ordered.append(m)
    return ordered

# Apply to your unique values array (replace `values` with your actual array/column)
practice_area_map = {}
for val in p2["Practice Areas"].unique():  # `values` = your array of unique strings
    practice_area_map[val] = standardize_practice_areas(val)

# Apply to the actual column
p2['Practice Areas (Standardized)'] = p2['Practice Areas'].apply(standardize_practice_areas)

# Convert list to comma-separated string
p2['Practice Areas (Standardized)'] = p2['Practice Areas (Standardized)'].apply(
    lambda x: ', '.join(x) if isinstance(x, list) else x
)

# Flag unmatched rows BEFORE overwriting (use the list output, empty list = no match)
no_match = p2[p2['Practice Areas (Standardized)'].apply(len) == 0]['Practice Areas'].unique()
if len(no_match) > 0:
    print(f"WARNING - {len(no_match)} values matched no category:")
    for v in no_match:
        print(f"  - {v}")

        
p2['Practice Areas'] = p2["Practice Areas (Standardized)"]
```

```python
# known client base might include companies that are restructured or liquidated or something like that,
# or companies that are unclear whether they are not commodity trader or maritime shipping,
# or companies that are not located in singapore, or companies with hq abroad, or companies not listed in geographic scope of the project
# source verification are inconsistent
p2['Known Client Base'] = p2['Known Client Base'].str.replace(r'\s+,\s*', ', ', regex=True)

# turn every client into a list of strings
def string_to_listOfStr(clients):
    if not isinstance(clients, str) or not clients.strip():  # <- guard for NaN/empty, unrelated to URLs
        return clients
    else:
        final_clients = []
        clients = re.split(r',\s*(?![^()]*\))', clients) 
        # split commas that are not inside (). , followed by \s* >=0 space and the condition to NOT split (?![]) condition where any character [^()]* follows by \)
        # means .split only if the comma does not meet the condition. condition to not meet is [^()]*\)
        for name in clients:
            final_clients.append(name.strip()) #name.strip() removes the first whitespace
        return final_clients
    
# p2["Known Client Base"] = p2["Known Client Base"].apply(string_to_listOfStr)
p2["Known Client Base"] = p2["Known Client Base"].apply(
    lambda x: ", ".join(x) if isinstance(x, list) else x
)

p2['Known Client Base']
```

```python
# check 'Notable Cases / Disputes'

# co. ltd or co, ltd or co., ltd to co ltd. natixis, singapore branch becomes natixis singapore branch.
# ltd. or ltd, to ltd
# pte. ot pte, to pte
# duncan, to duncan and
p2['Notable Cases / Disputes'] = p2['Notable Cases / Disputes'].str.replace(r'\b(co|ltd|pte|inc|corp)[.,]+', r'\1', regex=True, case=False)
p2['Notable Cases / Disputes'] = p2['Notable Cases / Disputes'].str.replace(r',\s*(singapore branch)', r' \1', regex=True, case=False)
p2['Notable Cases / Disputes'] = p2['Notable Cases / Disputes'].str.replace(r'\bDuncan,\s*', r'Duncan and ', regex=True, case=False)

# turn every case into an item inside a list
def string_to_listOfStr(cases):
    if (
        not isinstance(cases, str) or not cases.strip()
    ):  # <- guard for NaN/empty, unrelated to URLs
        return cases
    else:
        final_cases = []
        cases = re.split(r",\s*(?![^()]*\))", cases)
        # split commas that are not inside (). , followed by \s* >=0 space and the condition to NOT split (?![]) condition where any character [^()]* follows by \)
        # means .split only if the comma does not meet the condition. condition to not meet is [^()]*\)
        for name in cases:
            final_cases.append(
                name.strip()
            )  # name.strip() removes the first whitespace
        return final_cases

# p2['Notable Cases / Disputes'] = p2['Notable Cases / Disputes'].apply(string_to_listOfStr)
p2["Notable Cases / Disputes"] = p2["Notable Cases / Disputes"].apply(
    lambda x: ", ".join(x) if isinstance(x, list) else x
)
p2["Notable Cases / Disputes"]
# case label name (e.g. The {text}, The {text}) etc
# actual case name {text} v {text} etc. (many columns might appear even though it belongs to one {text} ex:COSCO Shipping Specialized Carriers Co, Ltd v PT OKI Pulp & Paper Mills and others [2023] SGHC 149)
# some case name {text} (Comm) etc
# some are literal descriptions (e.g. "Precedent setting FOSFA sale of goods arbitration (switzerland)", "Appealing the Refined Sugar Associations Arbitration Award (UK)") etc
# some case name {text} advises or advised or advising the {text}
# some case name are In re: {text}
# commas inside parenthesis are not to be split
# some case name is Re {text} ex: Re Brightoil Petroleum (S’pore) Pte Ltd [2022] SGHC 35
# some case name is {text} ex: $60 Million African Oil Project Arbitration (LCIA London)
# some case name is Case involving {text}
# some case name is Representing {text} ex: Representing Japanese ship owner Masao Mizobuchi (perplexity)
# some case name is acting for {text}
# some case naem is engaged by {text}
# some case name si advising on {text}
# in some case is In the matter of {text} or In re {text} or Re {text}
# some case name is dispute over {text} or dispute between {text}
# some case name is just {text} ex: PDV Marina Wind-Up (CV Shipping Liquidation)
```
